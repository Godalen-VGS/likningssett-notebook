# 🔴 E03: Likningssett med to ukjente

Et lineært likningssett med to ukjente kan skrives på formen

$$\begin{aligned} a_1 x + b_1 y &= c_1 \\ a_2 x + b_2 y &= c_2 \end{aligned}$$

Geometrisk er hver likning en rett linje, og løsningen er skjæringspunktet. To linjer kan

- skjære hverandre i **ett punkt** (én løsning),
- være **parallelle** (ingen løsning), eller
- være **den samme linja** (uendelig mange løsninger).

> Behold funksjonsnavnene og parameterne nøyaktig slik oppgaven sier. Det er det som gjør at selvtesten og den automatiske rettingen kan sjekke svarene dine.
